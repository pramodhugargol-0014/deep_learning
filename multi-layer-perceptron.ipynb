```python
import numpy as np
import pandas as pd

class MLP():
    def __init__(self, learning_rate):
        self.learning_rate = learning_rate
        # 4 inputs -> 3 neurons
        self.W1 = np.zeros((3, 4))
        self.b1 = np.zeros((3, 1))

        # 3 inputs -> 2 neurons
        self.W2 = np.zeros((2,3))
        self.b2 = np.zeros((2,1))

         # 2 inputs -> 1 neuron
        self.W3 = np.zeros((1,2))
        self.b3 = np.zeros((1,1))

    def sigmoid(self, z):
        return 1 / (1 + np.exp(-z))

    def forward(self, x):
        # Input -> Hidden Layer 1
        z1 = self.W1 @ x + self.b1
        h1 = self.sigmoid(z1)

        # Hidden Layer 1 -> Hidden Layer 2
        z2 = self.W2 @ h1 + self.b2
        h2 = self.sigmoid(z2)

        # Hidden Layer 2 -> Output
        z3 = self.W3 @ h2 + self.b3
        output = self.sigmoid(z3)
        return output
```
